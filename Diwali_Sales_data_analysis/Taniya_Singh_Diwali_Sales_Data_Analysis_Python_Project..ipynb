# Diwali Sales Analysis
######     Introduction:
This project analyzes Diwali sales data to understand customer purchasing behavior based on gender, age, marital status, occupation, state, and product category. The objective is to identify key trends that can help businesses improve marketing strategies and maximize festive season revenue.


###### #Load data
###### #Cleaning



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
%matplotlib inline

# Load dataset
df = pd.read_csv("Diwali Sales Data.csv", encoding='unicode_escape')

# Basic info
df.shape
df.info()
df.head()
df.shape
df.describe()

# Drop unnecessary columns
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)

# Check missing values
df.isnull().sum()

# Drop null values
df.dropna(inplace=True)

df.shape
```

##### Cleaning Summary:
- Removed unused columns (Status and unnamed1).
- Removed rows with missing values.
- Converted Amount column to integer type.
- Dataset is now clean and ready for analysis.

### Gender-wise Sales Analysis

```python
sales_gender = df.groupby('Gender')['Amount'].sum().sort_values(ascending=False)

sales_gender.plot(kind='bar')
plt.title("Total Sales by Gender")
plt.ylabel("Amount")
plt.show()
```

##### Observations:

Female customers contribute a higher total sales amount compared to male customers.

This indicates that women are more active shoppers during the Diwali season.

Festive marketing campaigns can focus more on female buyers to increase revenue.

The spending pattern shows strong purchasing behavior among women.

### Analysis of Total Sales Based on Age Group

```python
sales_age = df.groupby('Age Group')['Amount'].sum().sort_values(ascending=False)

sales_age.plot(kind='bar')
plt.title("Total Sales by Age Group")
plt.ylabel("Amount")
plt.show()
```

##### Observations:

The age group 26–35 generates the highest total sales.

Customers in the 18–25 and 36–45 age groups also contribute significantly.

Very young (0–17) and older age groups show comparatively lower spending.

This suggests that young adults and working professionals are the primary target customers during festive seasons.

### Marital Status-wise Sales Analysis

```python
sales_marital = df.groupby('Marital_Status')['Amount'].sum()

sales_marital.plot(kind='bar')
plt.title("Sales by Marital Status")
plt.show()
```

##### Observations:

- Married customers contribute more to total sales compared to unmarried customers.
- This indicates that married individuals tend to spend more during the festive season.
- Family responsibilities and gift purchasing may influence higher spending.
- Targeted festive offers for families can help increase overall sales revenue.

### State-wise Sales Analysis (Top 10 States)

```python
state_sales = df.groupby('State')['Amount'].sum().sort_values(ascending=False).head(10)

state_sales.plot(kind='bar')
plt.title("Top 10 States by Sales")
plt.show()
```

##### Observations:

- Maharashtra and Uttar Pradesh generate the highest sales revenue.
- These states show strong customer purchasing activity during Diwali.
- Northern and Western regions contribute significantly to overall sales.
- Businesses should focus more marketing efforts and festive offers in high-performing states to maximize revenue.

### Occupation-wise Sales Analysis

```python
occupation_sales = df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False)

occupation_sales.plot(kind='bar')
plt.title("Sales by Occupation")
plt.show()
```

##### Observations:

- Customers working in IT, Healthcare, and Aviation sectors show higher purchasing power.
- Working professionals contribute significantly to festive sales revenue.
- Occupations with stable income tend to spend more during Diwali.
- Marketing campaigns targeting salaried professionals can increase festive sales.

### Product Category-wise Sales Analysis

```python
product_sales = df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False)

product_sales.plot(kind='bar')
plt.title("Sales by Product Category")
plt.show()
```

### Observations:

- Clothing and Food categories generate the highest sales revenue.
- Customers prefer purchasing apparel and festive items during Diwali.
- Product categories related to lifestyle and celebrations perform better.
- Businesses should stock and promote high-demand categories during festive seasons.

##### Conclusion  :
- Female customers aged 26–35 are the primary buyers.
- Married individuals spend more during Diwali.
- Maharashtra and Uttar Pradesh generate highest revenue.
- IT and Healthcare professionals show strong purchasing power.
- Clothing and Food are top-performing categories.
